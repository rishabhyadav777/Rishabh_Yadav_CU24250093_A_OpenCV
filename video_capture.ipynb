```python
import cv2

cap = cv2.VideoCapture(0)

recording = False
out = None

while True:
    ret, frame = cap.read()

    if not ret:
        break

    cv2.imshow("Camera", frame)

    key = cv2.waitKey(1) & 0xFF

    # Press P to capture picture
    if key == ord('p'):
        cv2.imwrite("picture.jpg", frame)
        print("Picture saved!")

    # Press R to start/stop recording
    elif key == ord('r'):

        if not recording:
            fourcc = cv2.VideoWriter_fourcc(*'XVID')

            out = cv2.VideoWriter(
                "video.avi",
                fourcc,
                20.0,
                (frame.shape[1], frame.shape[0])
            )

            recording = True
            print("Recording started!")

        else:
            recording = False
            out.release()
            out = None
            print("Recording stopped!")

    # Write frames while recording
    if recording:
        out.write(frame)

    # Press Q to quit
    elif key == ord('q'):
        break

cap.release()

if out is not None:
    out.release()

cv2.destroyAllWindows()
```
